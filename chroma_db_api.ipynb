```python
collection = client.create_collection(name="my_collection")
```

```python
resume1 = 'My name is Navanit. My skills are Python, HTML DJango with 4 years of eaxperice'
resume2 = 'My name is Aisha. My skills are Java, Spring Boot, and MySQL with 5 years of experience.'
resume3 = 'My name is Rahul. My skills are JavaScript, React, and Node.js with 3 years of experience.re'
resume4 = 'My name is Priya. My skills are C#, .NET Core, and Azure with 6 years of experience.'
resume5 = 'My name is Daniel. My skills are PHP, Laravel, and Vue.js with 4.5 years of experience.'

collection.add(documents=[resume1,resume2,resume3,resume4,resume5], ids=['1','2','3','4','5'])
```
